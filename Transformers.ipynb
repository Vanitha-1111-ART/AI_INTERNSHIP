```python
from transformers import pipeline
classifier=pipeline("sentiment-analysis")
classifier("I Love You")
```

```python
classifier("sorryyyy")
```

```python
classifier("i didn't play worstly")
```

```python
classifier("this is not my worst day")
```

```python
classifier1=pipeline("zero-shot-classification")
classifier1("i really happy to win the competition",candidate_labels=["education","sports","politics"])
```

```python
generator=pipeline("text-generation")
generator("hey i am very ")
```

```python
generator=pipeline("text-generation",model="sapientinc/HRM-Text-1B")
generator("how was the work",max_length=40,num_return_sequences=2)
```

```python
fillmask =pipeline("fill-mask")
```

```python
fillmask("i am very <mask> to see ",top_k=3)
```

```python
NER=pipeline("ner")
```
