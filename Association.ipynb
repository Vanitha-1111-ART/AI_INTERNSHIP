```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
d=pd.read_csv('/content/Market_Basket_Optimisation.csv',header=None)
d.head()
```

```python
tr=[]
for i in range(0,len(d)):
  tr.append([str(d.values[i,j]) for j in range(0,20) if str(d.values[i,j])!='nan'])
```

```python
!pip install apyori
```

```python
from apyori import apriori
rules=apriori(transactions=tr,min_support=0.003,min_confidence=0.2,min_lift=3,min_length=2,max_length=2)
results=list(rules)
results
```

```python
import pandas as pd

rows = []

for item in results:

    support = item.support

    for stat in item.ordered_statistics:

        rows.append({
            "LHS": ", ".join(list(stat.items_base)),
            "RHS": ", ".join(list(stat.items_add)),
            "Support": support,
            "Confidence": stat.confidence,
            "Lift": stat.lift
        })

df = pd.DataFrame(rows)

print(df)
```

```python
import pandas as pd
from pyECLAT import ECLAT

# dataset load
df = pd.read_csv("/content/Market_Basket_Optimisation.csv", header=None)
eclat = ECLAT(data=df)

indices, support = eclat.fit(min_support=0.03,
                             min_combination=1,
                             max_combination=3)

print(support)
```

```python
import pandas as pd
r=[]
for key,value in support.items():
  r.append(
      {
        "Items": key,
        "Support":value
      }
  )
df=pd.DataFrame(r)
print(df)
```
