```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn.datasets
from sklearn.model_selection import train_test_split
```

```python
breast_cancer_dataset =sklearn.datasets.load_breast_cancer()
```

```python
print(breast_cancer_dataset)
```

```python
df=pd.DataFrame(breast_cancer_dataset.data,columns=breast_cancer_dataset.feature_names)
```

```python
df
```

```python
df['labels']=breast_cancer_dataset.target
df
```

```python
df.isnull().sum()
```

```python
df['labels'].value_counts()
```

```python
x=df.drop('labels',axis=1)
y=df[['labels']]
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=1)
```

```python
from sklearn.preprocessing import StandardScaler
model = StandardScaler()
```

```python
x_scaled = model.fit_transform(x_train)
y_scaled = model.fit_transform(y_train)
```

```python
model.fit_transform(x_test)
```

TENSERFLOW  KERAS

```python
import tensorflow as tf
from tensorflow import  keras
```

```python
models=keras.Sequential([keras.layers.Flatten(input_shape=(30,)),
                         keras.layers.Dense(20,activation='relu'),
                         keras.layers.Dense(2,activation='sigmoid')
                         ])
```

```python
models.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])
```

```python
history= models.fit(x_scaled ,y_train,validation_split=0.1,epochs=30)
```

```python
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model accuracy')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.legend(['training data','validation'],loc='lower right')
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model loss')
plt.xlabel('epoch')
plt.ylabel('loss')
plt.legend(['training data','validation'],loc='lower right')
```
