```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers,datasets,Model
from tensorflow.keras.layers import Input,LSTM,Dense,Bidirectional
```

```python
mnist=keras.datasets.mnist
```

```python
(train_images,train_labels),(test_images,test_labels)=mnist.load_data()
```

```python
train_images,test_images=train_images/255.0,test_images/255.0
```

```python
import matplotlib.pyplot as plt
plt.imshow(train_images[234])
```

```python
from numpy import shape
input_shape=(28,28)
inputs=Input(shape=input_shape)
rnn_layer=Bidirectional(LSTM(128))(inputs)
output=Dense(10,activation="softmax")(rnn_layer)
model=Model(inputs=inputs,outputs=output)
```

```python
model.compile(optimizer="adam",loss="sparse_categorical_crossentropy",metrics=['accuracy'])
```

```python
h=model.fit(train_images,train_labels,validation_split=0.2,epochs=5)
```

```python
y_pre=model.predict(test_images)
```

```python
import numpy as np
yp=np.argmax(y_pre,axis=1)
yp[234]
```
